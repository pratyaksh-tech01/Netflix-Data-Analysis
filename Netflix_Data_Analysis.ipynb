```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("netflix_titles.csv")

print("Rows and Columns:", df.shape)

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['type'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)

plt.title("Movies vs TV Shows")
plt.ylabel("")
plt.show()
```

```python
df['country'].value_counts().head(10).plot(
    kind='bar',
    figsize=(10,5),
    color='skyblue'
)

plt.title("Top 10 Countries on Netflix")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```

```python
df['release_year'].value_counts().head(15).plot(
    kind='bar',
    figsize=(12,5),
    color='orange'
)

plt.title("Top Release Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```
